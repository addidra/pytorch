### shape errors when matmul
1. inner dim must match
* (3,2) @ (3,2) won't work
* (2,3) @ (3,2) will work
* (4,2) @ (2,5) will work
2. The result mat has the shape of outer dim
* (4,2) @ (2,5) = (4,5)

```python
(torch.rand(4,2) @ torch.rand(2,5)).shape
```

```python
# shape for mat mul
tensor_A = torch.tensor([[1,2],
                         [3,4],
                         [5,6]])
tensor_B = torch.tensor([[7,10],
                         [8,11],
                         [9,12]])

torch.mm(tensor_A, tensor_B)
```

to fix our tensor shape issues we can manipulate of shape of one tensor using transpose


```python
print(torch.mm(tensor_A, tensor_B.T))
print(torch.mm(tensor_A.T, tensor_B))
```

## Reshaping, stacking, squeezing, unsqueezing tensors
* Reshaping - reshpae the input tensor to defined shape
* View -  return a view of input tensor of certain shape but keep the smae memory as original tensor
* Stacking - combine multiple tensors on top of each other (vstack) or side by side (hstack)
* Squeeze - removes all '1' dimensions from a tensor
* unsqueeze - add a '1' dimension to a target tensor
* permute - return a view of the input with dimensions permuted (swapper) in a certain way

```python
x = torch.arange(1.,13.)
x, x.shape
```

```python
x.reshape(4,3)
```

```python
z = x.view(4,3)
```

```python
z[:,0:1] = 1
z,x
```

```python
z_stacked = torch.stack([z,z,z], dim=2)
z_stacked,z_stacked.shape
```

```python
x_reshaped = x.reshape(1,12)
x_squeezed = x_reshaped.squeeze()
x_reshaped, x_reshaped.shape, x_squeezed, x_squeezed.shape
```

```python
print(f"Prev target: {x_squeezed}")
print(f"Prev shape: {x_squeezed.size()}")

x_unsqueezed = x_squeezed.unsqueeze(dim=1).unsqueeze(1).unsqueeze(0)
print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
print(f"New shape: {x_unsqueezed.squeeze()}")
```

```python
# torch.permute = rearange the shape and order of the tensor
# permute is the view of orignal, meaning changes in permuted would change x_orignal

x_original = torch.rand(size=(224,224,3))
x_permuted = torch.rand(x_original.permute(2,0,1).shape)
x_original.shape, x_permuted.shape
x_permuted[0,0,0] = 1
x_original[0,0,0], x_permuted[0,0,0]
```

## pytorch tensors and numpy
* data in numpy, want in pytorch tensor -> `torch.from_nmpy(ndarray)`
* pytorch tensor -> numpy -> `torch.Tensor.numpy()`

```python
import torch
import numpy as np

array = np.arange(1.0,8.0)
tensor = torch.from_numpy(array)
back_to_np = tensor.numpy()
array, tensor,back_to_np
# they dont share memory like permute and view does
```

## Reproducbility (trying to take random out of random)
`start with random -> tensor operation -> update random num -> loop`
reduce the randomness in nn and pytorch comes the concept of random seed.
essentially what the random seed does is flavour the randomness
lets make random but reproducable

```python
import torch

RANDOM_SEED = 0
torch.manual_seed(RANDOM_SEED)
random_tensor_A = torch.rand(3,6)

torch.manual_seed(RANDOM_SEED)
random_tensor_B = torch.rand(3,6)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_B == random_tensor_A)
```
